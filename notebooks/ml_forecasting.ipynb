```python
import pandas as pd
import pandas as pd

# Tampilkan semua kolom tanpa terpotong
pd.set_option('display.max_columns', None)
```

```python
df = pd.read_excel("05. Database RP May 2025 - AC REGISTER.xlsx")
df
```

```python
# Gunakan baris ke-1 sebagai header
df.columns = df.iloc[0]  # set baris ke-1 sebagai header
df = df.drop([0])     # hapus baris pertama (NaN) dan baris header lama

# Reset index
df = df.reset_index(drop=True)
df = df.iloc[:,1:]
df
```

```python
for column in df.columns:
    nunique = df[column].nunique()
    unique = list(df[column].unique())
    print(column)
    print(df[column])
    print(f"nunique: {nunique}")
    if nunique <= 50:
        print(f"unique: {unique}")
    print("")
```
