```python
import pandas as pd
```

```python
df = pd.read_excel("../05. Database RP May 2025 - AC REGISTER.xlsx", sheet_name="Raw", skiprows=1)
df = df.iloc[:, 1:]
df.head()
```

```python
cols = ["YEAR", "PERIODE", "AIRCRAFT TYPE", "FLIGHT NUMBER_CITYPAIR", "SERVICE TYPE", "SUB-SERVICE", 
        "ROUNDTRIPROUTE", "FLIGHT ROUTE", "AC REG", "Region"]
```

```python
df1 = df.loc[:, cols]
df1.drop_duplicates(inplace=True)
```

```python
unique_dict = {col: df1[col].unique().tolist() for col in df1.columns}

# Find the maximum length among all unique value lists
max_len = max(len(v) for v in unique_dict.values())

# Pad lists with None to ensure all have the same length
for col in unique_dict:
    unique_dict[col].extend([None] * (max_len - len(unique_dict[col])))

# Create the new DataFrame
unique_df = pd.DataFrame(unique_dict)
unique_df
```

```python
unique_df.to_csv("df_filter1.csv", index=False)
```
