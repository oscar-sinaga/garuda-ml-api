## Remove zero

```python
df = df[df['VARIABLE MAINTENANCE'] >= 0].copy()
```

```python
df = df[df['VARIABLE MAINTENANCE'] >= 0].copy()

df1 = df[
    (df['FUEL BURN (IN LITER)'] != 0) &
    (df['FLIGHT HOURS'] != 0) &
    (df['VARIABLE MAINTENANCE'] != 0)
].copy()
```

## Finding Corelation

```python
df_num = df.select_dtypes(include=["number"])

# Misal ingin korelasi semua numerik dengan VARIABLE MAINTENANCE
corr = df_num.corrwith(df_num['VARIABLE MAINTENANCE']).sort_values(ascending=False).dropna()

print(corr.head(30))
```

## Group VM per PK Register per Periode

```python
df_group = df1.groupby(["AC REG", "PERIODE"]).agg({
    "VARIABLE MAINTENANCE": "sum",
    "BLOCK HOURS": "sum",
    "FLIGHT HOURS": "sum",
    "FUEL BURN (IN LITER)": "sum",
    "FUEL AIRCRAFT": "sum",
    "ASK (000)": "sum",
    "ATK PASSENGER (000)": "sum",
    "ATK (000)": "sum",
    "LEASE AIRCRAFT": "mean",
    "CABIN CREW TRAVEL": "sum",
    "COCKPIT CREW TRAVEL": "sum",
    "CABIN CREW PERSON": "mean",
    "ROUNDTRIPROUTE": "nunique", 
    "FLIGHT ROUTE": "nunique", 
    "SERVICE TYPE": "first", 
    "AIRCRAFT TYPE": "first",
}).reset_index()
```

```python
# Filter periode Januari
df_group_jan = df_group[df_group['PERIODE'].str.upper() == 'JAN']

# Tampilkan hasil
print(df_group_jan)
```

```python
print(df_group)
```

```python
import numpy as np

df_group['FH_per_BH'] = df_group['FLIGHT HOURS'] / df_group['BLOCK HOURS'].replace(0, np.nan)
df_group['Fuel_per_FH'] = df_group['FUEL BURN (IN LITER)'] / df_group['FLIGHT HOURS'].replace(0, np.nan)
df_group['ATK_per_ASK'] = df_group['ATK (000)'] / df_group['ASK (000)'].replace(0, np.nan)

df_group = df_group.fillna(0)
```

## Input Selected Features

```python
selected_features = [
    'BLOCK HOURS',
    'FLIGHT HOURS',
    'FUEL BURN (IN LITER)',
    'ASK (000)',
    'ATK PASSENGER (000)',
    'ATK (000)',
    'LEASE AIRCRAFT',
    'CABIN CREW TRAVEL',
    'FUEL AIRCRAFT',
    'COCKPIT CREW TRAVEL',
    'CABIN CREW PERSON',
    'ROUNDTRIPROUTE', 
    'AIRCRAFT TYPE',
    'SERVICE TYPE', 
    'AC REG',
    'FLIGHT ROUTE',
    'PERIODE',
]
```

```python
selected_features += ["FH_per_BH", "Fuel_per_FH", "ATK_per_ASK"]
```

## SPlit Data

```python
X = df_group[selected_features].copy()
y = df_group['VARIABLE MAINTENANCE'].copy()
```

```python
print (X)
```

```python
# from sklearn.model_selection import train_test_split


# X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
```

```python
from sklearn.model_selection import GroupShuffleSplit

splitter = GroupShuffleSplit(test_size=0.2, random_state=42)
train_idx, test_idx = next(splitter.split(df_group, groups=df_group["AC REG"]))

X_train = X.iloc[train_idx]
X_test = X.iloc[test_idx]
y_train = y.iloc[train_idx]
y_test = y.iloc[test_idx]
```

```python
categorical_cols = ['AC REG','PERIODE','ROUNDTRIPROUTE','AIRCRAFT TYPE','SERVICE TYPE','FLIGHT ROUTE']
numeric_cols = list(set(selected_features) - set(categorical_cols))
```

```python
print(numeric_cols)
```

## Encoder

```python
from sklearn.preprocessing import OneHotEncoder


encoder = OneHotEncoder(handle_unknown='ignore', sparse_output=False)
encoder.fit(X_train[categorical_cols])
```

```python
X_train_encoded = encoder.transform(X_train[categorical_cols])
X_test_encoded = encoder.transform(X_test[categorical_cols])
```

```python
X_train_encoded = encoder.transform(X_train[categorical_cols])
X_test_encoded = encoder.transform(X_test[categorical_cols])
encoded_cols = encoder.get_feature_names_out(categorical_cols)

X_train_encoded = pd.DataFrame(X_train_encoded, columns=encoded_cols, index=X_train.index)
X_test_encoded = pd.DataFrame(X_test_encoded, columns=encoded_cols, index=X_test.index)
```

```python
X_train_encoded = encoder.transform(X_train[categorical_cols])
X_test_encoded = encoder.transform(X_test[categorical_cols])
encoded_cols = encoder.get_feature_names_out(categorical_cols)

X_train_encoded = pd.DataFrame(X_train_encoded, columns=encoded_cols, index=X_train.index)
X_test_encoded = pd.DataFrame(X_test_encoded, columns=encoded_cols, index=X_test.index)
X_train_final = pd.concat([X_train[numeric_cols], X_train_encoded], axis=1)
X_test_final = pd.concat([X_test[numeric_cols], X_test_encoded], axis=1)
```

## PREDIKSI

```python
from xgboost import XGBRegressor

bst = XGBRegressor(
    n_estimators=500, 
    max_depth=10, 
    learning_rate=0.05, 
    gamma=0.1, 
    objective="reg:squarederror"
)


bst.fit(X_train_final, y_train)
```

```python
y_pred = bst.predict(X_test_final)
```

```python
from sklearn.metrics import mean_absolute_percentage_error

y_pred = bst.predict(X_test_final)
mean_absolute_percentage_error(y_test, y_pred)
```
