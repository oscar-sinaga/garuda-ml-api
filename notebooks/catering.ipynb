```python
df1 = df[(df['PASSENGER CARRIED'] != 0) & (df['BLOCK HOURS']!=0) & 
        (df['ON BOARD SERVICE']!=0) & (df['CATERING']!=0)].copy()
len(df1)
```

```python
SELECTED_FEATURES = [
                     'PASSENGER CARRIED', 
                     'ATK (000)', 'ATK PASSENGER (000)', 'ASK (000)', 'ASK (000) Y CLASS',
                     'FLIGHT ROUTE', 'SERVICE TYPE', 'Region']

df1[SELECTED_FEATURES]
```

```python
X = df1[SELECTED_FEATURES].copy()
y = df1['CATERING'].copy()
```

```python
CATEGORICAL_COLS = ['SERVICE TYPE', 'FLIGHT ROUTE', 'Region']
NUMERIC_COLS = list(set(SELECTED_FEATURES) - set(CATEGORICAL_COLS))
NUMERIC_COLS
```

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_percentage_error, mean_squared_error
from sklearn.preprocessing import OneHotEncoder
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=10)
```

```python
encoder = OneHotEncoder(handle_unknown='ignore', sparse_output=False)
encoder.fit(X_train[CATEGORICAL_COLS])
```

```python
X_train_encoded = encoder.transform(X_train[CATEGORICAL_COLS])
X_test_encoded = encoder.transform(X_test[CATEGORICAL_COLS])
```

```python
encoded_cols = encoder.get_feature_names_out(CATEGORICAL_COLS)

X_train_encoded = pd.DataFrame(X_train_encoded, columns=encoded_cols, index=X_train.index)
X_test_encoded = pd.DataFrame(X_test_encoded, columns=encoded_cols, index=X_test.index)
```

```python
X_train_final = pd.concat([X_train[NUMERIC_COLS], X_train_encoded], axis=1)
X_test_final = pd.concat([X_test[NUMERIC_COLS], X_test_encoded], axis=1)
```

```python
from xgboost import XGBRegressor
```

```python
model = XGBRegressor(n_estimators=2000, learning_rate=0.5, objective="reg:squarederror")
model.fit(X_train_final, y_train)
```

```python
y_pred = model.predict(X_test_final)
```

```python
mean_absolute_percentage_error(y_test, y_pred)
```
