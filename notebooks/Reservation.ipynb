### Remove zeroes

```python
df1 = df[(df['CARGO CARRIED']!=0) & (df['FREIGHT CARRIED']!=0) & (df['PASSENGER CARRIED'] != 0) &
        (df['BLOCK HOURS']!=0) & (df['RESERVATION']!=0)].copy()
len(df1)
```

```python
SELECTED_FEATURES = ['PASSENGER CARRIED', 'PASSENGER CARRIED Y CLASS', 'PASSENGER CARRIED C CLASS', 'CARGO CARRIED',
                     'RPK (000)', 'RPK (000) Y CLASS', 'SEAT OFFERED', 'SEAT OFFERED Y CLASS',
                    'FLIGHT ROUTE', 'SERVICE TYPE', 'AIRCRAFT TYPE', 'AIRCRAFT TYPE GROUPING', 'Region']

df1[SELECTED_FEATURES]
```

```python
X = df1[SELECTED_FEATURES].copy()
y = df1['RESERVATION'].copy()
```

```python
CATEGORICAL_COLS = ['SERVICE TYPE', 'FLIGHT ROUTE', 'AIRCRAFT TYPE', 'AIRCRAFT TYPE GROUPING', 'Region']
NUMERIC_COLS = list(set(SELECTED_FEATURES) - set(CATEGORICAL_COLS))
NUMERIC_COLS
```

### Split into training and testing

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_percentage_error, mean_squared_error
from sklearn.preprocessing import OneHotEncoder
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
```

```python
encoder = OneHotEncoder(handle_unknown='ignore', sparse_output=False)
encoder.fit(X_train[CATEGORICAL_COLS])
```

```python
X_train_encoded = encoder.transform(X_train[CATEGORICAL_COLS])
X_test_encoded = encoder.transform(X_test[CATEGORICAL_COLS])
```

```python
encoded_cols = encoder.get_feature_names_out(CATEGORICAL_COLS)

X_train_encoded = pd.DataFrame(X_train_encoded, columns=encoded_cols, index=X_train.index)
X_test_encoded = pd.DataFrame(X_test_encoded, columns=encoded_cols, index=X_test.index)
```

```python
X_train_final = pd.concat([X_train[NUMERIC_COLS], X_train_encoded], axis=1)
X_test_final = pd.concat([X_test[NUMERIC_COLS], X_test_encoded], axis=1)
```

```python
from xgboost import XGBRegressor
```

## XGBoost

```python
model = XGBRegressor(n_estimators=1500, learning_rate=0.01, objective="reg:squarederror")
model.fit(X_train_final, y_train)
```

```python
y_pred = model.predict(X_test_final)
```

```python
mean_absolute_percentage_error(y_test, y_pred)
```

## LightGBM

```python
from lightgbm import LGBMRegressor
```

```python
model = LGBMRegressor(
    n_estimators=1500,
    learning_rate=0.01,
    num_leaves=255,
    max_depth=-1,
    # feature_fraction=0.9,
    # bagging_fraction=0.9,
    # bagging_freq=1,
    min_data_in_leaf=20
)

model.fit(X_train_final, y_train)
```

```python
y_pred = model.predict(X_test_final)
```

```python
mean_absolute_percentage_error(y_test, y_pred)
```
