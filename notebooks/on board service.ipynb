```python
df1 = df[(df['PASSENGER CARRIED'] != 0) & (df['BLOCK HOURS']!=0) & 
        (df['ON BOARD SERVICE']!=0) & (df['CATERING']!=0)].copy()
len(df1)
```

```python
SELECTED_FEATURES = ['PASSENGER CARRIED',
                    'ATK PASSENGER (000)', 'ASK (000)', 'ASK (000) Y CLASS', 'ATK (000)',
                    'CABIN CREW PERSON', 'COCKPIT CREW PERSON'
                    ]

df1[SELECTED_FEATURES]

# performed worse when combined with these categorical columns:  'FLIGHT ROUTE', 'SERVICE TYPE', 'Region'
```

```python
X = df1[SELECTED_FEATURES].copy()
y = df1['ON BOARD SERVICE'].copy()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_percentage_error, mean_squared_error
from sklearn.preprocessing import OneHotEncoder
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=10)
```

## XGBoost

```python
from xgboost import XGBRegressor
```

```python
model = XGBRegressor(n_estimators=2000, learning_rate=0.5, objective="reg:squarederror")
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
mean_absolute_percentage_error(y_test, y_pred)
```
